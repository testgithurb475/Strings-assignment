### 61. A store offers a 10% discount on a product that costs 1500. Write a Python expression to calculate the discounted price using the appropriate arithmetic operator. Ask the user to enter the product price and discount rate.

```python
price = float(input("Enter the product price: "))
discount_rate = float(input("Enter the discount rate (e.g., 10 for 10%): "))
discounted_price = price - (price * (discount_rate / 100))
print("Discounted price:", discounted_price)
```

### 62. You have some apples. Ask the user to enter the total number of apples they have and the number of apples they give to a friend. Write a Python program to calculate how many apples are left. Which operator will you use to find the remaining apples?

```python
total_apples = int(input("Enter total number of apples: "))
given_apples = int(input("Enter number of apples given to a friend: "))
apples_left = total_apples - given_apples
print("Apples left:", apples_left)
```

### 63. Ask the user to enter the ages of two friends. Write a Python program to check who is older. Which comparison operator will you use to find out who is older?

```python
age1 = int(input("Enter age of first friend: "))
age2 = int(input("Enter age of second friend: "))
if age1 > age2:
    print("The first friend is older.")
elif age2 > age1:
    print("The second friend is older.")
else:
    print("Both friends are of the same age.")
```
